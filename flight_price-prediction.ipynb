```python
x = df2.drop('price',axis=1)
y = df2['price']
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test= train_test_split(x, y, test_size=0.2, random_state=42)
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.fit_transform(x_test)
```

DECISION TREE REGRESSOR

```python
from sklearn.tree import DecisionTreeRegressor
dtree = DecisionTreeRegressor(random_state=0)
dtree.fit(x_train,y_train)
```

```python
y_pred = dtree.predict(x_test)
```

```python
from sklearn.metrics import r2_score
print("Decision Tree Regression R2 Score: ", r2_score(y_test, y_pred))
```

```python
from sklearn import metrics
import math
mae = metrics.mean_absolute_error(y_test, y_pred)
mse = metrics.mean_squared_error(y_test, y_pred)
r2 = metrics.r2_score(y_test, y_pred)
rmse = math.sqrt(mse)

print(f'MAE is: {mae}')
print(f'MSE is: {mse}')
print(f'R2 score is: {r2}')
print(f'RMSE score is: {rmse}')
```

Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(random_state=0)
rf.fit(x_train,y_train)
```

```python
from sklearn import metrics
import math
y_pred = rf.predict(x_test)
mae = metrics.mean_absolute_error(y_test, y_pred)
mse = metrics.mean_squared_error(y_test, y_pred)
r2 = metrics.r2_score(y_test, y_pred)
rmse = math.sqrt(mse)

print('MAE is {}'.format(mae))
print('MSE is {}'.format(mse))
print('R2 score is {}'.format(r2))
print('RMSE score is {}'.format(rmse))
```

ADABOOST REGRESSOR

```python
from sklearn.ensemble import AdaBoostRegressor

ab = AdaBoostRegressor(random_state=0)
ab.fit(x_train, y_train)
```

```python
import math

y_pred_ab = ab.predict(x_test)

mae_ab = metrics.mean_absolute_error(y_test, y_pred_ab)
mse_ab = metrics.mean_squared_error(y_test, y_pred_ab)
r2_ab = metrics.r2_score(y_test, y_pred_ab)
rmse_ab = math.sqrt(mse_ab)

print('AdaBoost Regressor Evaluation:')
print(f'MAE is {mae_ab}')
print(f'MSE is {mse_ab}')
print(f'R2 score is {r2_ab}')
print(f'RMSE score is {rmse_ab}')
```
